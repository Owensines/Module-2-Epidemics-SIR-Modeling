```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the data
data = pd.read_csv('../Data/mystery_virus_daily_active_counts_RELEASE#1.csv',
                   parse_dates=['date'],
                   header=0,
                   index_col=None)

# Rename long column for easier use
data = data.rename(columns={'active reported daily cases': 'active'})

# Plot active infections over time
plt.figure()
plt.plot(data['date'], data['active'])

plt.xlabel('Date')
plt.ylabel('Active Infections')
plt.title('Mystery Virus: Active Infections Over Time')

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
